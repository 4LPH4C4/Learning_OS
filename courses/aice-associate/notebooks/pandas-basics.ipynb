# pandas 기초 실습

Concept → Example → Guided Exercise → Independent Exercise → Answer Check 순서로 진행한다. 작은 인라인 데이터셋만 사용한다.

## Concept
`DataFrame`은 표 데이터다. 필터는 조건에 맞는 행을 고르고, `groupby`는 그룹별 요약을 만든다.

```python
import pandas as pd

sales = pd.DataFrame({
    'product': ['책', '책', '펜', '펜', '노트'],
    'quantity': [2, 1, 3, 2, 4],
    'price': [12000, 12000, 1500, 1500, 3000]
})
sales['revenue'] = sales['quantity'] * sales['price']
sales
```

## Example
매출이 10,000원 이상인 행과 상품별 총매출을 확인한다.

```python
high_sales = sales[sales['revenue'] >= 10000]
by_product = sales.groupby('product', as_index=False)['revenue'].sum()
high_sales, by_product
```

## Guided Exercise
`quantity`가 3 이상인 행만 골라 `guided`에 저장하고, 그 행들의 평균 단가를 계산해라.

```python
guided = sales[sales['quantity'] >= 3]
guided_average_price = guided['price'].mean()
guided, guided_average_price
```

## Independent Exercise
상품별 평균 수량을 `average_quantity`에 저장해라. 가장 평균 수량이 높은 상품도 찾아라.

```python
average_quantity = sales.groupby('product', as_index=False)['quantity'].mean()
top_product = average_quantity.loc[average_quantity['quantity'].idxmax(), 'product']
average_quantity, top_product
```

## Answer Check
아래 셀을 실행해 독립 연습의 결과를 확인한다. 상품별 평균 수량은 노트 4.0, 펜 2.5, 책 1.5이며 가장 높은 상품은 노트다.

```python
expected = {'노트': 4.0, '펜': 2.5, '책': 1.5}
actual = dict(zip(average_quantity['product'], average_quantity['quantity']))
assert actual == expected, (actual, expected)
assert top_product == '노트'
print('정답 확인 완료')
```
